# Creator Daily Stats — Silver Transformation

- Bronze → Silver
- Inspect Bronze creator daily stats
- Exclude Auto Loader rescue column
- Standardize data types
- Write Silver Delta table
- Validate Silver data

## 1. Inspect Bronze data

```python
%sql
-- Preview Bronze data
SELECT *
FROM youtube_content_intelligence.bronze.brz_creator_daily_stats
LIMIT 5;
```

```python
%sql
-- Inspect Bronze schema
DESCRIBE TABLE youtube_content_intelligence.bronze.brz_creator_daily_stats;
```

## 2. Transform Bronze data

```python
from pyspark.sql.functions import col, to_date

# Load Bronze table
df_bronze = spark.table("youtube_content_intelligence.bronze.brz_creator_daily_stats")

# Exclude Auto Loader rescue column and standardize data types
df_silver = df_bronze.select(
    col("channel_id"),
    to_date("collection_date").alias("collection_date"),
    col("subscriber_count").cast("long").alias("subscriber_count"),
    col("total_view_count").cast("long").alias("total_view_count"),
    col("video_count").cast("int").alias("video_count"),
    col("ingested_at")
)
```

```python
display(df_silver)
```

```python
df_silver.printSchema()
```

## 3. Write to Silver

```python
df_silver.write.format("delta").mode("overwrite").saveAsTable(
    "youtube_content_intelligence.silver.slv_creator_daily_stats"
)
```

## 4. Validate Silver table

```python
%sql
-- Preview Silver data
SELECT *
FROM youtube_content_intelligence.silver.slv_creator_daily_stats;
```

```python
%sql
-- Check row count
SELECT COUNT(*) AS row_count
FROM youtube_content_intelligence.silver.slv_creator_daily_stats;
```

```python
%sql
-- Check row count
SELECT COUNT(*) AS row_count
FROM youtube_content_intelligence.silver.slv_creator_daily_stats;
```

```python
%sql
-- Check duplicate daily records
SELECT channel_id, collection_date, COUNT(*) AS count
FROM youtube_content_intelligence.silver.slv_creator_daily_stats
GROUP BY channel_id, collection_date
HAVING COUNT(*) > 1;
```
